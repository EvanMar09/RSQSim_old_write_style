In this getting started notebook we will go over some simple manipulation of RSQSim outputs:

For the old-write output style, RSQsim will output several different files:

OUTPUTS: 

eqs.filename.out : This is the file that has all the cataloged events

filename.eList : This file has all the events which slip,stress,etc are recorded at each element (Binary)

filename.pList : This file contains the patches that participated (Slipped) in each event from the .eList file (Binary)

filename.dList : This file contains the slip for each element during an event. (Binary)

filename.tList : This file contains the time for which each element started slipping during an event. (Binary)

filename.dtauList : This file contains the shear stress for each element during an event. (Binary)

OTHER FILES NEEDED:

filename.flt : This is the original fault file you used to run the simulation. This will be important to have in your working directory

### Important Note: This assumes you have already ran the RSQSim simulations and have the old-write-style output

### Step 1: Import dependencies and set up file pathes


```python
## Packages and dependencies 
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import math
from mpl_toolkits.mplot3d import Axes3D
from scipy.stats import invgauss
from scipy.stats import weibull_min
import pyvista as pv

### May add in my plotting functions later ##
```

```python
indir = "/Users/emarschall/Desktop/RSQSim_stuff/step_over/step_over1_outputs/" # directory containing the RSQSim outputs: will need to modify 
outdir = indir #Directory where you want to save plots and what not 
filename = "step_over_test"
event_file = "eqs."+filename+".out" # this is the file which conatins all the events
fault_file = "Step_over.flt" # Name of your fault file 
## These files are Binary ##
e_file = filename+".eList" # Event number
p_file = filename+".pList" # list of patches in each event
s_file = filename+".dList" # list of slip of each element in event (M)
t_file = filename+".tList" # list of slip-on-set-time of each element in event (seconds)
tau_file = filename+".dtauList" # list of shear stress change  of each element in event (MPa)

## I like to make a folder to keep all the figures
folder_name = outdir + "figures/"
try:
    os.makedirs(folder_name, exist_ok=True)
    print(f"Folder '{folder_name}' exists.")
except OSError as e:
    print(f"Error creating folder: {e}")
    
```

### Step 2: Extract the data from the files
eqs.filename.out has columns for :  
t0 : origin time in seconds  
M0 : seismic moment in N-m  
Mw moment magnitude : (Mw = (2/3)(log10(M0)− 9.1))  
x, y, z  : hypocentral coordinates in meters  
area : total area which ruptured in m2  
dt : duration in seconds  
nuc patch  : (if flagged in the initial RSSQSim run)  

```python
# Read in the event file
## must first find the number of lines to skip (These are often the input parameters used in the run)
starting_index =[]
beginning_string = "%%% end input files"
with open(indir+event_file, 'r') as file:
    for line_num, line in enumerate(file, 0):
        if beginning_string in line:
            print(f"Found '{beginning_string}' on line {line_num}: {line.strip()}")
            starting_index.append(line_num)
file.close()

col_names = ["t","Mo","Mw","nuc_x","nuc_y","nuc_z","area","duration","nuc_element"]
df_events_raw = pd.read_csv(indir+event_file, sep = " ", header =None, skiprows = starting_index[0]+1, names =col_names)

# Read in the fault file
fault = pd.read_csv(indir+fault_file, sep = " ", header =None)


#Read in the binary Files
data_events = np.fromfile(indir+e_file, dtype=np.int32) # Events
data_patches = np.fromfile(indir+p_file, dtype=np.int32) # Patches in each rupture 
data_slip = np.fromfile(indir+s_file, dtype='<f8')
data_time = np.fromfile(indir+t_file, dtype='<f8')
data_delta_tau = np.fromfile(indir+tau_file, dtype='<f8')






```

### Step 3: Lets do some basic post processing and plotting

RSQSim, like other cycle simulations, requires some start up-spin up time to ensure we aren't overly dependent on the initial conditions. So I usually like to filter out the dataframe for events only after that spin up time.  

However, for this example we only ran a very short RSQSim run for practical purposes. So in the example we will set spin-up time to 0 years. PLEASE NOTE THIS IS NOT STANDARD PRACTICE. 

Also, the time in RSQSim is in seconds so we should convert to years to make it easier to understand the results.



```python
# Filter the dataset 
spin_up_time = 0 #years
sec2yr = 3600*24*365.25
df_events_raw.t = df_events_raw.t/(sec2yr) #change from seconds to years 
#Remove the spin-up time events
df_spin_up = df_events_raw[(df_events_raw.t>= spin_up_time)]


```

#### Lets start basic and plot the magnitude frequency distribution  
For this we will use the dataframe with the spin up time removed and bin the magnitude by 0.25

```python
interval = 0.25 # This is the interval of the magnitude bins
max_event = np.ceil((max(df_spin_up.Mw.values))) #Find the maximum Magnitude and round up to get the upper bin limit
min_event = round((min(df_spin_up.Mw.values))) #Find the minimum Magnitude and round down to get the lower bin limit
bins = np.arange(min_event, max_event, interval) # bins for the histogram
# define the histogram
hist = np.histogram(df_spin_up.Mw.values,bins=bins,)
# Now make the plot 
fig, ax = plt.subplots()
ax.hist(df_spin_up.Mw.values,bins=bins,color = "skyblue",alpha = 0.8)
ax.set_xlabel("Magnitude (Mw)")
ax.set_yscale('log', base=10)
ax.set_ylabel("LogN # of Events")
ax.set_title("RSQSim Magnitude_Frequency Distribution")

 
fig.savefig(folder_name +"Magnitude-Frequency Plot")


```

#### Lets do something more advanced, lets plot some hazard curves of M > 6 earthquakes on 1 of the fault segments in the model  

For this we will fit a Brownian passage time (BPT) and a Weibull ditribution (Wei) for events that are greater than M > 6 nucleating on seg1 

This part has been inspired by work Mike Oskin's has been doing at UC Davis.


```python
M_threshold = 6 # Only want events equal to or large than this 
fault_string =  "seg1" # Name of the faault segment you are interested in
indices_fault = fault[fault[12].values == fault_string].index # Which elements in the model belong to the fault of interest 
t_clust = 0 # This is the year clustering parameter used in the iner_event times below (Keep reading) 
#In paleoseimicity it may be hard to differnetiate an event from multiple evnts if the interevent time is short.
# Furthermore, because RSQSim is not fully dynamic, sometimes it will have clustered events right nest to each other, these may just represent a single event if the full physics were implemented
# See Kroll et al., 2023 where they compare with FaultMod (t_clust ~ 4 years



# We will filter to only get events that are above our Mag threshold, after spin up time and that nucleate on our segment of interest 
## one note ## this will read the whole catalog for all times after spin_up, to reduce the total time frame you can define a different max_time

max_time = 1e100 # Max_time for looking at events, I recommend if the dataset is very large that you change this value
filtered_df = df_events_raw[(df_events_raw.Mw >= M_threshold)&(df_events_raw.t>= spin_up_time)&(df_events_raw.t< max_time)&(df_events_raw["nuc_element"].isin(indices_fault))]

# Lets calculate the inter_event times #### *Read the note, for simplicity t_clust her is set to zero 

inter_event_times = np.diff(filtered_df.t.values)
inter_event_times[np.where(inter_event_times> t_clust)]
inter_event_times.sort() # sort by ascending order
inter_event_times_rank_order = np.linspace(0,1,len(inter_event_times)) # This will be used to plot against the cumulative hazard curves later
time_frame = int(np.floor(max((inter_event_times)))+25) # Usually I like to have the time frame be a couple years larger than the greatest recurrance interval
t = np.linspace(0,time_frame,time_frame+1)

# Things needed for fitting to the distributions 
mean_inter_event_time = np.mean(inter_event_times) # mean 
std_inter_event_time = np.std(inter_event_times) # standard deviation
alpha = std_inter_event_time / mean_inter_event_time # aperiodicty

### Brownian Passage Time (BPT) ###
bpt_model = invgauss(alpha**2, scale=mean_inter_event_time)
cdf_bpt = bpt_model.cdf(t) #F(T) # probaility it has happened # Cumulative
pdf_bpt = bpt_model.pdf(t) #f(t) 
sf_bpt = bpt_model.sf(t) #1-F(t) # Probility it hasn't happened at time t

#Hazard function h(t) = f(t) / (1 - F(t)) Brownian Passage Time
hazard_rate_bpt = pdf_bpt / sf_bpt # This will give you the annual probability of an event happening 


### Weibull Distribution ###
k_wei, loc_wei, lambda_wei = weibull_min.fit(inter_event_times, floc=0)
dist_wei = weibull_min(k_wei, scale=lambda_wei) 
hazard_rate_wei = dist_wei.pdf(t) / dist_wei.sf(t) # Anuual Probability
cdf_wei = dist_wei.cdf(t) # Cumulative 


## Ok Lets Plot 

fig,[ax1,ax2] = plt.subplots(1, 2, figsize=(10, 4))
fig.suptitle("Hazard curves for "+fault_string) 

# Plot the annual rate first 
ax1.plot(t, hazard_rate_bpt, color='orange', lw=2, label='BPT Annual Probability ')
ax1.plot(t, hazard_rate_wei, color='blue', lw=2, label='Wei Annual Probability: k = 'f"{k_wei:.2f}")
ax1.axvline(mean_inter_event_time, color='black', linestyle='--', label='Mean Recurrance')
ax1.set_xlabel('Years Since Last Earthquake')
ax1.set_ylabel('Annual Probability of Occurrence')
ax1.set_title('Time-Dependent Earthquake Hazard Curve on ' + fault_string)
ax1.grid(True, alpha=0.3)
ax1.legend(loc = "lower right", fontsize=8)
plt.tight_layout()

# Plot the Cumulative Hazard
ax2.plot(inter_event_times,inter_event_times_rank_order,"ko",markersize=3, alpha = 0.5, label = "Inter-event Times")
ax2.plot(t, cdf_bpt, color='orange', lw=2, label='BPT Cumulative Probability ')
ax2.plot(t, cdf_wei, color='blue', lw=2, label='Wei Cumulative Probability: k = 'f"{k_wei:.2f}")
ax2.axvline(mean_inter_event_time, color='black', linestyle='--', label='Mean Recurrance')
ax2.set_xlabel('Years Since Last Earthquake')
ax2.set_ylabel('Probability of Occurrence')
ax2.set_title('Time-Dependent Annual Hazard: ' + fault_string)
ax2.grid(True, alpha=0.3)
ax2.legend(loc = "lower right", fontsize=8)
plt.tight_layout()

fig.savefig(folder_name +"Magnitude-Frequency Plot")


```

### Step 4: Lets try some 3D plots for individual events
This section will go over some basics of plotting slip for individual events in a 3d plot  
This will require pyvista

```python
# All the mesh info is sotred in the .flt file which we called in as a dataframe earlier and named fault
# Fault has the columns which correspond to the vertices of the faces in the first 9 columns  x1,y1,z1,x2,y2,z,2,x3,y3,z3
print(fault.iloc[:, :9].values[0])
```

 #### Lets first just visualize our fault system

```python
# Lets make a numpy array of the vertices
arr_vertices = np.array(fault.iloc[:, :9].values) # get the location of each vertice which amkes up the triangle
vertices = arr_vertices.reshape(-1, 3) #reshape into a long list of vertices

# Now need to construct the triangles from the vertices
num_triangles = arr_vertices.shape[0]
face_connectivity = np.arange(num_triangles * 3).reshape(-1, 3)
faces = np.column_stack([np.full(num_triangles, 3), face_connectivity]).flatten()

# Now we make a pyvisat mesh from our vertices and faces
mesh = pv.PolyData(vertices, faces)

## Plotting the meshed surface 
plotter = pv.Plotter()
plotter.add_mesh(mesh, color='lightgray', opacity=0.2, show_edges=False)
plotter.add_axes()
plotter.show_grid()
plotter.set_background("white") 
plotter.screenshot(folder_name +"fault3d")
plotter.show() # I do not have trame installed, so this might show an error/user warning but it will still plot
```

#### Now we will extract the slip for a single event and plot it 

```python
# Ok, now it can be a bit tricky and long-winded to get the slip data for an event

# Lets start by selecting an event, maybe the largest event in the catalog after spin-up time has been removed
Target_Event_index = df_spin_up[df_spin_up.Mw.values == np.max(df_spin_up.Mw.values)].index +1
#Important note, the datframe is zero indexed, but the data files called in earlier begin at 1 not zero, so we must add one to the index from the datframe

# Now lets find the patches which rupture during that event
# Keep in mind the data from the binary files is 1 indexed while the eqs and fault files read in earlier will be zero indexed
indices_patches = np.where(data_events == Target_Event_index[0])[0]
patches = data_patches[indices_patches] -1 #This is a list of the patches which ruptured during the event, -1 to get back to zero indexing we used in our faces mesh
slip = data_slip[indices_patches]

# Now lets add this to our mesh
full_slip_array = np.zeros(num_triangles) #Make all slip zero
full_slip_array[patches] = slip # assign the slip values to patches that slipped
mesh.cell_data['Slip (m)'] = full_slip_array # Assign to mesh

        
### And Now We Plot ###
plotter = pv.Plotter()
plotter.add_mesh(mesh, color='gray', opacity=0.1, show_edges=False) # just the fault background
rupture_mesh = mesh.threshold(0.001, scalars='Slip (m)')

plotter.add_mesh(rupture_mesh, 
                 cmap='jet', 
                 scalars='Slip (m)', 
                 show_edges=False, 
                 edge_color='white', 
                 line_width=0.5)

plotter.add_scalar_bar(title="Slip (m)")
plotter.add_axes()
plotter.show_grid()
plotter.set_background("white") 
plotter.screenshot(folder_name +"ruptured_fault3d")
plotter.show()

## This process can be used for the the other data from the binary files as well (stress_drop, ruptureTime)
```
